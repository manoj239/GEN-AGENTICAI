```python
import pandas as pd
try:
    df = pd.read_csv(r'C:\Users\manojkumar.borra\Desktop\AllAI\7.Python\heart.csv')
    print(df.head())
except FileNotFoundError:
    print("Error: File not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
print(df.shape) #Gives the number of rows and columns in the DataFrame
print(df.info()) #Gives a concise summary of the DataFrame, including the index dtype and columns, non-null values and memory usage
print(df.describe()) #Gives statistical summary of numerical columns
print(df.columns) #Gives the column names
print(df.dtypes) #Gives the data types of each column
print(df.head(10)) #Gives the first 10 rows of the DataFrame
print(df.tail()) #Gives the last 5 rows of the DataFrame
print(df.tail(10)) #Gives the last 10 rows of the DataFrame
print(df.isnull().sum()) #Gives the count of missing values in each column
print(df['age'].value_counts()) #Gives the count of unique values in the 'age' column
print(df['age'].unique()) #Gives the unique values in the 'age' column
print(df['age'].nunique()) #Gives the number of unique values in the 'age' column
print(df['age'].mean()) #Gives the mean of the 'age' column
print(df['age'].median()) #Gives the median of the 'age' column
print(type(df)) #Gives the type of the object, which is a DataFrame
```

```python
print("\n Number of zeros per column:")
for col in df.columns:
    zero_count = (df[col] == 0).sum()
print(f"{col}: {zero_count}")
"""Above code will print the number of zeros in each column of the DataFrame. It iterates through each column, counts the number of zeros, and prints the result."""
```

```python
df.rename(columns={'gender': 'sex'},inplace=True)
print(df.head()) #Rename the column name gender to sex
print(df[df['Disease'] == 1]) #Gives the rows where the value of the column 'Disease' is 1
print(df[(df['Disease'] == 0) & (df['Sex'] == 'male')]) #Gives the rows where the value of the column 'Disease' is 0 and the Sex is male
```

```python
print(df.loc[df['age'] > 50]) #Gives all the rows where the age is greater than 50
print(df.loc[:, 'Age']) #Gives all the rows for the column 'Age'
print(df.loc[:, ['Age', 'Sex']]) #Gives all the rows for the columns 'Age' and 'Sex'
print(df.loc[df['age'] > 50, ['age', 'sex']]) #Gives all the rows where the age is greater than 50 and only the columns 'age' and 'sex'
print(df.iloc[[1, 2]]) #Gives the first two rows  and all the columns in the DataFrame
print(df.iloc[[1, 2], [0, 1]]) #Gives the first two rows and the first two columns in the DataFrame
print(df.iloc[1:3, 0:2]) #Gives the rows from index 1 to 2 and the columns from index 0 to 1 in the DataFrame
print(df.iloc[1:3, [0, 2]]) #Gives the rows from index 1 to 2 and the columns at index 0 and 2 in the DataFrame
print(df.iloc[:,[2,4,6]]) #Gives all the rows and the columns at index 2, 4 and 6 in the DataFrame
print(df.iloc[1:3, [0, 2]]) #Gives the rows from index 1 to 2 and the columns at index 0 and 2 in the DataFrame
```

```python
print(df.groupby('Sex')[['age','Disease']].mean()) #Gives the mean of the columns 'age' and 'Disease' grouped by the column 'Sex'
print(df.groupby('Gender')['MaxHeartRate'].sum().reset_index()) #Gives the sum of the column 'MaxHeartRate' grouped by the column 'Gender'
print(df.groupby('Gender')['MaxHeartRate'].max().reset_index())
print(df.groupby('Gender')['MaxHeartRate'].min().reset_index())
print(df.groupby('Gender')['MaxHeartRate'].agg(['count', 'mean', 'min', 'max']).reset_index())
print(df.groupby('column')['numeric_column'].agg(['count','sum','mean','min','max'])) #Pattern to get count, sum, mean, min and max of a numeric column grouped by a categorical column
```
